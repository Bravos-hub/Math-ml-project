# Final-dataset EDA
This notebook is descriptive only. It reads the authoritative multi-crop panel and reports distributions, missingness, correlations, crop/season/year structure, and target-scale imbalance.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
p = Path('data/processed/final_multi_crop_subregion_season_year.csv')
df = pd.read_csv(p)
df.describe(include='all').T
```

```python
df.groupby(['crop','season','year']).size().rename('n').reset_index()
```

```python
df.isna().mean().sort_values(ascending=False).head(20)
df.groupby('crop')['yield_tons_ha'].agg(['count','mean','median','std','min','max']).sort_values('mean')
```

```python
numeric = df.select_dtypes('number')
numeric.corr(numeric_only=True)['yield_tons_ha'].sort_values()
```
